```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,0:4],df.iloc[:,-1],test_size=0.2,random_state=2)
X_train.head()
```

```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder

# OneHotEncoder banana
ohe = OneHotEncoder(drop='first', dtype=np.int32)

# Only categorical columns choose karein
X_train_cat = X_train[['fuel', 'owner']]
X_test_cat = X_test[['fuel', 'owner']]

# Fit only on train (important!)
X_train_new = ohe.fit_transform(X_train_cat)

# Test ko same encoder se transform
X_test_new = ohe.transform(X_test_cat)

print("Shape of encoded X_train:")
print(X_train_new.shape)

# Numerical + categorical ko combine karna
X_train_final = np.hstack((
    X_train[['brand', 'km_driven']].values,   # numeric / string columns
    X_train_new                                # one-hot encoded columns
))

print("Final X_train shape:", X_train_final.shape)
print(X_train_final)
```
